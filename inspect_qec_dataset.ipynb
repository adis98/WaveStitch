```python
from data_utils import PreprocessorOrdinal

preprocessor = PreprocessorOrdinal(name = "QECParityZZ")
train_ids = preprocessor.train_indices
test_ids= preprocessor.test_indices

original_df = preprocessor.df_orig
train_df = original_df.iloc[train_ids]
test_df = original_df.iloc[test_ids]

train_shots = sorted(train_df["shot_id"].unique())
test_shots = sorted(test_df["shot_id"].unique())

print("Amount of train shots:", len(train_shots), ", max train shots:", max(train_shots), ", min train shots:", min(train_shots))
print("Amount of test shots:", len(test_shots), ", max test shots:", max(test_shots), ", min test shots:", min(test_shots))
```
